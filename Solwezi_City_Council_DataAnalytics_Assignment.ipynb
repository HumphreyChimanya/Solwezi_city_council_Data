## Data Visualization

### Subtask:
Create a series of appropriate data visualizations (e.g., bar charts, line graphs, scatter plots) to illustrate key findings, trends, and patterns from the analysis. Ensure all plots are clearly labeled with titles, axis labels, and legends for better understanding.


Visualize the distribution of 'sector' using a bar chart showing the count of projects. We will use `integrated_df['sector'].value_counts()` to get the data and `seaborn.barplot` to create the visualization.



```python
plt.figure(figsize=(12, 7))
sns.barplot(x=integrated_df['sector'].value_counts().index, y=integrated_df['sector'].value_counts().values)
plt.title('Distribution of Projects by Sector')
plt.xlabel('Sector')
plt.ylabel('Number of Projects')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Filter out 'Unknown' or totals for better comparative plotting
ward_plot_df = ward_summary[~ward_summary.index.isin(['Unknown', 'Total Amount:', 'All', 'Kapijipanga', 'Tumvwanganai'])]

plt.figure(figsize=(14, 6))
x = range(len(ward_plot_df))
width = 0.35

plt.bar([pos - width/2 for pos in x], ward_plot_df['total_amount_requested'], width, label='Total Requested', color='skyblue')
plt.bar([pos + width/2 for pos in x], ward_plot_df['total_amount_recommended'], width, label='Total Recommended', color='salmon')

plt.xlabel('Ward')
plt.ylabel('Amount (ZMK / Kwacha)')
plt.title('Total Requested vs. Recommended Funding by Ward')
plt.xticks(x, ward_plot_df.index, rotation=45, ha='right')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Group Type comparison
group_plot_df = group_type_summary[~group_type_summary.index.isin(['Unknown', 'Commuunity', 'Communty'])]

plt.figure(figsize=(10, 6))
sns.barplot(x=group_plot_df.index, y=group_plot_df['funding_ratio'], palette='viridis')
plt.title('Funding Ratio (Recommended / Requested) by Group Type')
plt.xlabel('Group Type')
plt.ylabel('Funding Success Ratio')
plt.ylim(0, 1.0)
plt.tight_layout()
plt.show()
```

## Summary of Findings & Conclusion

Based on the analysis of the Solwezi Municipal Council's 2025 CDF tables, we have uncovered several key insights:

1. **Agriculture Dominance**: Out of the categorized sectors, **Agriculture** holds the largest portion of project counts (53 projects) and total funding requests (~2.49M Kwacha) with a healthy funding success ratio of ~57.9%.
2. **Group Dynamics**: **Community** groups applied for the most projects and received a higher portion of their requested budgets (~58.1%) compared to **Women** (~52.3%) and **Youth** (~33.9%). This indicates potential space for policy intervention to support youth entrepreneurship.
3. **Ward Distributions**: Wards like **Kyawama**, **Tuvwanganai**, **Kyalalankuba**, and **Kimasala** are highly active in proposing projects. Kyawama, in particular, secured an impressive funding ratio of ~90% of its requested capital.
