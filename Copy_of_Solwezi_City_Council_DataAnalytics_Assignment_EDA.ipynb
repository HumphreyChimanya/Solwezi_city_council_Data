We need to analyze the unique values and their counts for key categorical columns (`sector`, `ward`, `zone`, `group_type`, `venture_type`, `comment`, `project_description`, `group_name`, `contact_person`) to understand their distribution and fill any remaining missing categorical values with 'Unknown'.



```python
print("--- Integrated DataFrame Head ---")
display(integrated_df.head())

print("\n--- Integrated DataFrame Info ---")
integrated_df.info()

print("\n--- Integrated DataFrame Descriptive Statistics ---")
display(integrated_df.describe(include='all'))

print("\n--- Integrated DataFrame Missing Values ---")
missing_values = integrated_df.isnull().sum()
missing_percentage = (missing_values / len(integrated_df)) * 100
missing_info = pd.DataFrame({
    'Missing Count': missing_values,
    'Missing Percentage': missing_percentage
})
display(missing_info[missing_info['Missing Count'] > 0].sort_values(by='Missing Percentage', ascending=False))

print("\n--- Analyzing Categorical Columns ---")
# Columns to analyze unique values and their counts
categorical_cols_to_analyze = ['sector', 'ward', 'zone', 'group_type', 'venture_type', 'comment', 'project_description', 'group_name', 'contact_person']

for col in categorical_cols_to_analyze:
    if col in integrated_df.columns:
        print(f"\nUnique values and their counts for '{col}':")
        display(integrated_df[col].value_counts(dropna=False))

# Fill remaining missing categorical values with 'Unknown' where appropriate
for col in categorical_cols_to_analyze:
    if col in integrated_df.columns and integrated_df[col].dtype == 'object':
        integrated_df[col] = integrated_df[col].fillna('Unknown')

print("\n--- Missing values after filling categorical NaNs ---")
missing_values_after_fill = integrated_df.isnull().sum()
missing_percentage_after_fill = (missing_values_after_fill / len(integrated_df)) * 100
missing_info_after_fill = pd.DataFrame({
    'Missing Count': missing_values_after_fill,
    'Missing Percentage': missing_percentage_after_fill
})
display(missing_info_after_fill[missing_info_after_fill['Missing Count'] > 0].sort_values(by='Missing Percentage', ascending=False))
```

**Reasoning**:
I need to fill the remaining missing numerical values in the `integrated_df` with their respective medians to ensure data completeness for further analysis.



```python
print("--- Filling remaining missing numerical values with median in integrated_df ---")

# Identify numerical columns with missing values that should be filled with median
numerical_cols_to_fill = [
    'engineers_estimates',
    'allocated_amount_cdf',
    'amount_requested',
    'amount_recommended_cdf'
]

for col in numerical_cols_to_fill:
    if col in integrated_df.columns and integrated_df[col].isnull().any():
        median_value = integrated_df[col].median()
        integrated_df[col] = integrated_df[col].fillna(median_value)
        print(f"Filled missing values in '{col}' with median: {median_value}")


print("\n--- Missing values after filling numerical NaNs ---")
missing_values_final = integrated_df.isnull().sum()
missing_percentage_final = (missing_values_final / len(integrated_df)) * 100
missing_info_final = pd.DataFrame({
    'Missing Count': missing_values_final,
    'Missing Percentage': missing_percentage_final
})
display(missing_info_final[missing_info_final['Missing Count'] > 0].sort_values(by='Missing Percentage', ascending=False))

print("\nCompleted filling numerical missing values in integrated_df.")
```

We need to analyze the distribution of key categorical variables to understand the types of projects, groups, and sectors involved, as this is a fundamental part of exploratory data analysis.



```python
print("\n--- Analyzing Distributions of Key Categorical Variables ---")

categorical_eda_cols = [
    'sector',
    'ward',
    'zone',
    'group_type',
    'venture_type'
]

for col in categorical_eda_cols:
    if col in integrated_df.columns:
        print(f"\nDistribution of '{col}':")
        display(integrated_df[col].value_counts(dropna=False))
```


To ensure the full distributions are displayed for each categorical variable, we will print the value counts for each selected column individually.



```python
print("\n--- Analyzing Distributions of Key Categorical Variables Individually ---")

categorical_eda_cols = [
    'sector',
    'ward',
    'zone',
    'group_type',
    'venture_type'
]

for col in categorical_eda_cols:
    if col in integrated_df.columns:
        print(f"\nDistribution of '{col}':")
        # Using to_string() to avoid truncation in output display
        print(integrated_df[col].value_counts(dropna=False).to_string())
```


The next step in EDA is to analyze the distributions of numerical variables. We will use histograms to visualize these distributions and descriptive statistics to summarize them.



```python
import matplotlib.pyplot as plt
import seaborn as sns

print("\n--- Analyzing Distributions of Key Numerical Variables ---")

numerical_eda_cols = [
    'engineers_estimates',
    'allocated_amount_cdf',
    'amount_requested',
    'amount_recommended_cdf'
]

plt.figure(figsize=(15, 10))
for i, col in enumerate(numerical_eda_cols):
    if col in integrated_df.columns:
        plt.subplot(2, 2, i + 1)
        sns.histplot(integrated_df[col], kde=True)
        plt.title(f'Distribution of {col}')
        plt.xlabel(col)
        plt.ylabel('Frequency')
plt.tight_layout()
plt.show()

print("\n--- Descriptive Statistics for Key Numerical Variables ---")
display(integrated_df[numerical_eda_cols].describe())
```

We need to analyze the average requested and recommended amounts by key categorical variables like `sector`, `ward`, and `group_type` to understand funding distributions and identify potential areas of interest or disparity. This will provide a quantitative view of the relationships between categorical and numerical data.



```python
print("\n--- Analyzing Average Amounts by Categorical Variables ---")

categorical_grouping_cols = [
    'sector',
    'ward',
    'group_type'
]

for col in categorical_grouping_cols:
    if col in integrated_df.columns:
        print(f"\nAverage Amount Requested and Recommended by '{col}':")
        grouped_data = integrated_df.groupby(col)[['amount_requested', 'amount_recommended_cdf']].mean().sort_values(by='amount_requested', ascending=False)
        print(grouped_data.to_string())
```

To analyze funding by venture type, We need to group the integrated DataFrame by the 'venture_type' column and then calculate the mean of 'amount_requested' and 'amount_recommended_cdf'.



```python
print("\n--- Average Amount Requested and Recommended by 'venture_type' ---")

grouped_by_venture_type = integrated_df.groupby('venture_type')[['amount_requested', 'amount_recommended_cdf']].mean().sort_values(by='amount_requested', ascending=False)
print(grouped_by_venture_type.to_string())
```

We will calculate the funding ratio (amount recommended / amount requested) for each venture type to assess the success rate of funding applications and identify ventures that are more likely to receive funding relative to their requests.



```python
print("\n--- Funding Ratio (Recommended/Requested) by 'venture_type' ---")
grouped_by_venture_type['funding_ratio'] = grouped_by_venture_type['amount_recommended_cdf'] / grouped_by_venture_type['amount_requested']
print(grouped_by_venture_type.sort_values(by='funding_ratio', ascending=False).to_string())
```

We will calculate and display the number of projects and the total amount requested and recommended for each ward to understand funding distribution across different geographical areas.



```python
print("\n--- Project Count and Total Funding by 'ward' ---")

ward_summary = integrated_df.groupby('ward').agg(
    project_count=('project_no', 'count'),
    total_amount_requested=('amount_requested', 'sum'),
    total_amount_recommended=('amount_recommended_cdf', 'sum')
).sort_values(by='total_amount_requested', ascending=False)
print(ward_summary.to_string())
```

We will calculate and display the number of projects and the total amount requested and recommended for each `group_type` to understand funding distribution among different types of groups.



```python
print("\n--- Project Count and Total Funding by 'group_type' ---")

group_type_summary = integrated_df.groupby('group_type').agg(
    project_count=('project_no', 'count'),
    total_amount_requested=('amount_requested', 'sum'),
    total_amount_recommended=('amount_recommended_cdf', 'sum')
).sort_values(by='total_amount_requested', ascending=False)
print(group_type_summary.to_string())
```

We will calculate and display the number of projects and the total amount requested and recommended for each `sector` to understand funding distribution across different types of sectors.



```python
print("\n--- Project Count and Total Funding by 'sector' ---")

sector_summary = integrated_df.groupby('sector').agg(
    project_count=('project_no', 'count'),
    total_amount_requested=('amount_requested', 'sum'),
    total_amount_recommended=('amount_recommended_cdf', 'sum')
).sort_values(by='total_amount_requested', ascending=False)
print(sector_summary.to_string())
```

We will calculate the funding ratio for each sector by dividing the 'total_amount_recommended' by the 'total_amount_requested' in the `sector_summary` DataFrame and then display the results.



```python
print("\n--- Funding Ratio (Recommended/Requested) by 'sector' ---")
sector_summary['funding_ratio'] = sector_summary['total_amount_recommended'] / sector_summary['total_amount_requested']
print(sector_summary.sort_values(by='funding_ratio', ascending=False).to_string())
```
