```python
import pandas as pd
import requests
from requests.packages.urllib3.exceptions import InsecureRequestWarning
import io

# Suppress only the single warning from urllib3
requests.packages.urllib3.disable_warnings(InsecureRequestWarning)

url = 'https://www.solwezicouncil.gov.zm/?page_id=11011'

try:
    # Use requests to get the HTML content, disabling SSL verification
    response = requests.get(url, verify=False)
    response.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)
    html_content = response.text

    # Read tables from the HTML content, wrapping it in StringIO to avoid FutureWarning
    tables = pd.read_html(io.StringIO(html_content))

    if tables:
        print(f"Found {len(tables)} table(s) on the page. Displaying the first one:")
        # Display the first table found
        display(tables[0])

        # iterate through 'tables' to see all of them
         for i, df_table in enumerate(tables):
             print(f"\nTable {i+1}:")
             display(df_table)
    else:
        print("No tables found on the specified URL.")
except requests.exceptions.RequestException as e:
    print(f"An error occurred while fetching the URL: {e}")
except Exception as e:
    print(f"An error occurred: {e}")
```

### Save tables to CSV files

save each of the extracted tables to a separate CSV file. This will make it easier to load and preprocess them in a different notebook.


```python
import os

output_dir = 'extracted_tables'
os.makedirs(output_dir, exist_ok=True)

for i, df_table in enumerate(tables):
    file_name = os.path.join(output_dir, f'table_{i+1}.csv')
    df_table.to_csv(file_name, index=False)
    print(f"Table {i+1} saved to {file_name}")

print(f"\nAll {len(tables)} tables have been saved to the '{output_dir}' directory.")
```
